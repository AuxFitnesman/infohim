# EDA (кратко)
Распределения таргетов, пропуски в свойствах, суммы долей по `scenario_id`.

```python
from pathlib import Path
import pandas as pd

root = Path.cwd().resolve().parent
train = pd.read_csv(root / "data" / "daimler_mixtures_train.csv")
props = pd.read_csv(root / "data" / "daimler_component_properties.csv")
c_delta = "Delta Kin. Viscosity KV100 - relative | - Daimler Oxidation Test (DOT), %"
c_eot = "Oxidation EOT | DIN 51453 Daimler Oxidation Test (DOT), A/cm"
y = train.drop_duplicates("scenario_id")[[c_delta, c_eot]]
display(y.describe())
s = train.groupby("scenario_id")["Массовая доля, %"].sum()
display(s.describe())
print("Пропуски в значениях свойств:", props["Значение показателя"].isna().mean())
```
