# Daimler DOT — инференс
Одна команда через `nbconvert --execute` или запуск этой ячейки: формируется `predictions.csv`.

```python
import subprocess
import sys
from pathlib import Path

root = Path.cwd().resolve()
cmd = [
    sys.executable,
    "-m",
    "src.predict",
    "--data-dir",
    str(root / "data"),
    "--artifacts-dir",
    str(root / "artifacts"),
    "--output",
    str(root / "predictions.csv"),
    "--device",
    "cpu",
]
print("Running:", " ".join(cmd))
subprocess.check_call(cmd, cwd=str(root))
```
